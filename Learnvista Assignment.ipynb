```python
#Write a Python program to find the sum of all elements in a list using loop.
#Input:- [10,20,30,40]
#Output:- 100

# Function to calculate the sum of all elements in a list
def sum_of_elements(lst):
    total = 0
    for num in lst:
        total += num
    return total

# Input list
numbers = [10, 20, 30, 40]

# print the result
result = sum_of_elements(numbers)
print("Sum of all elements:", result)
```

```python
#Write a Python program to find the multiplication of all elements in a list using loop.
#Input:- [10,20,30,40]
#Output:- 240000

# Function to calculate the product of all elements in a list
def multiply_elements(lst):
    product = 1   # Assume the first product is 1
    for num in lst:
        product *= num #"Product *= num" means (Product= Product*number){Short form in python}
    return product

# Input list
numbers = [10, 20, 30, 40]

# print the result
result = multiply_elements(numbers)
print("Multiplication of all elements:", result)
```

```python
#Write a Python program to find the largest number from a list using loop.
#Input:- [10,100,2321, 1,200,2]
#Output:- 2321

# Function to find the largest number in a list
def find_largest(lst):
    largest = lst[0]  # Assume the first element is the largest initially
    for num in lst:
        if num > largest:   # Compare each element with the current largest
            largest = num   # Update the largest if current number is bigger
    return largest  # Return to the largest number after completeing the loop

# Input list
numbers = [10, 100, 2321, 1, 200, 2]

# print the result
result = find_largest(numbers)
print("The largest number is:", result)
```

```python
#Write a Python program to find the smallest number from a list using loop.
#Input:- [10,100,2321, 1,200,2]
#Output:- 1

# Function to find the smallest number in a list
def find_smallest(lst):
    smallest = lst[0]  # Assume the first element is the smallest initially
    for num in lst:
        if num < smallest:  # Compare each element with the current smallest
            smallest = num  # Update the smallest if current number is smaller
    return smallest

# Input list
numbers = [10, 100, 2321, 1, 200, 2]

# Call the function and print the result
result = find_smallest(numbers)
print("The smallest number is:", result)
```

```python
#Write a Python program to count the number of strings having length more than 2 and are palindrome in a list using loop.
#Input:- ['ab', 'abc', 'aba', 'xyz', '1991']
#Output:- 2

def count_palindromes(lst):
    count = 0  # Initialize a counter
    for string in lst:
        if len(string) > 2 and string == string[::-1]:  # To Check the length and palindrome condition (To check if a string is a palindrome, we compare the string to its reverse. If both are the same, the string is a palindrome.)
            count += 1  # Increment the counter if both conditions are met
    return count

# Input list
strings = ['ab', 'abc', 'aba', 'xyz', '1991']

# Call the function and print the result
result = count_palindromes(strings)
print("Number of palindromes with length greater than 2:", result)
```
